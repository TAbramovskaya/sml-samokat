```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
orders = pd.read_excel("../data/orders.xlsx")
products = pd.read_excel("../data/products.xlsx")
sales = orders.merge(products, on="product_id")
```

#### ABC analysis at the subcategory (level2) level

```python
subcategories_summary = (
    sales
    .groupby('level2')[['quantity', 'price']]
    .sum()
    .reset_index()
    .rename(columns={'price': 'revenue'})
)
```
